# Volleyball Feeder Shape Exploration

This notebook inspects the MP-GCN volleyball feeder output shapes.


```
import os
import sys
from pprint import pprint

repo_root = os.path.abspath('..')
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

from src.dataset.graphs import Graph
from src.dataset.volleyball_feeder import Volleyball_Feeder

CONFIG = {
    'graph': 'coco_ball-12',
    'labeling': 'intra-inter',
    'inputs': 'JVBM',
    'input_dims': 2,
    'window': [10, 30],
    'ball': True,
    'person_id': list(range(12)),
    'inter_link': 'pairwise',
    'hop': 1,
    'dilation': 2,
}

root_folder = os.path.join(repo_root, 'data', 'volleyball')
phase = 'eval' if os.path.exists(os.path.join(root_folder, 'eval_data.npy')) else 'val'
print(f'Using split: {phase}')

graph = Graph(
    dataset='volleyball',
    graph=CONFIG['graph'],
    labeling=CONFIG['labeling'],
    inter_link=CONFIG['inter_link'],
    hop=CONFIG['hop'],
    dilation=CONFIG['dilation']
)

feeder = Volleyball_Feeder(
    phase=phase,
    graph=graph,
    root_folder=root_folder,
    inputs=CONFIG['inputs'],
    debug=False,
    ball=CONFIG['ball'],
    object_folder=root_folder,
    window=CONFIG['window'],
    person_id=CONFIG['person_id'],
    input_dims=CONFIG['input_dims']
)

shape = feeder.get_datashape()
print('Feeder data shape [I, C, T, V, M]:', shape)

sample, label, name = feeder[0]
print('Sample full tensor shape:', sample.shape)
for idx, branch in enumerate(CONFIG['inputs']):
    print(f'  Branch {idx} ({branch}) shape:', sample[idx].shape)
print('Sample label:', label)
print('Sample clip name:', name)
```
